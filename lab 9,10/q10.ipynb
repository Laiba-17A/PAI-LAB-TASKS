```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

data = {
    'Overall': [88, 91, 79, 84, 76, 90, 85, 73, 80, 92],
    'Potential': [90, 93, 82, 86, 78, 91, 87, 75, 82, 94],
    'Age': [28, 33, 22, 26, 30, 31, 25, 21, 27, 29],
    'IntReputation': [4, 5, 3, 3, 2, 5, 4, 2, 3, 5],
    'Value_M': [95, 120, 45, 70, 30, 110, 75, 25, 55, 130]
}

df = pd.DataFrame(data)

X_full = df[['Overall', 'Potential', 'Age', 'IntReputation']]
y = df['Value_M']

X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(X_full, y, test_size=0.25, random_state=14)

model_full = LinearRegression()
model_full.fit(X_train_f, y_train_f)
pred_full = model_full.predict(X_test_f)

impact_full = sorted(zip(X_full.columns, np.abs(model_full.coef_)), key=lambda x: -x[1])

X_no_pot = df[['Overall', 'Age', 'IntReputation']]
X_train_n, X_test_n, y_train_n, y_test_n = train_test_split(X_no_pot, y, test_size=0.25, random_state=14)

model_no_pot = LinearRegression()
model_no_pot.fit(X_train_n, y_train_n)
pred_no_pot = model_no_pot.predict(X_test_n)

print("Full Model")
print("R2:", r2_score(y_test_f, pred_full))
print("RMSE:", np.sqrt(mean_squared_error(y_test_f, pred_full)))
print("Most Influential Attribute:", impact_full[0])

print("\nWithout Potential")
print("R2:", r2_score(y_test_n, pred_no_pot))
print("RMSE:", np.sqrt(mean_squared_error(y_test_n, pred_no_pot)))
```
