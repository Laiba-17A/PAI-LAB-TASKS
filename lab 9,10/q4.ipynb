```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df_original = pd.DataFrame({
    "height": [150, 155, 160, 165, 170, 175, 180],
    "weight": [50, 55, 60, 63, 68, 72, 75]
})

X_orig = df_original[["height"]]
y_orig = df_original["weight"]
model_orig = LinearRegression()
model_orig.fit(X_orig, y_orig)
y_pred_orig = model_orig.predict(X_orig)

print("original dataset")
print("slope:", round(model_orig.coef_[0],2))
print("intercept:", round(model_orig.intercept_,2))
print("r2:", round(r2_score(y_orig, y_pred_orig),3))

plt.scatter(df_original["height"], df_original["weight"])
plt.plot(df_original["height"], y_pred_orig, color='red')
plt.xlabel("height (cm)")
plt.ylabel("weight (kg)")
plt.title("original dataset")
plt.show()

df_outlier = pd.DataFrame({
    "height": [150, 155, 160, 165, 170, 175, 180, 190],
    "weight": [50, 55, 60, 63, 68, 72, 75, 60]
})

X_out = df_outlier[["height"]]
y_out = df_outlier["weight"]
model_out = LinearRegression()
model_out.fit(X_out, y_out)
y_pred_out = model_out.predict(X_out)

print("dataset with outlier")
print("slope:", round(model_out.coef_[0],2))
print("intercept:", round(model_out.intercept_,2))
print("r2:", round(r2_score(y_out, y_pred_out),3))

plt.scatter(df_outlier["height"], df_outlier["weight"])
plt.plot(df_outlier["height"], y_pred_out, color='red')
plt.xlabel("height (cm)")
plt.ylabel("weight (kg)")
plt.title("dataset with outlier")
plt.show()
```

```python
print("The outlier pulls the regression line towards itself, which changes the slope and intercept.")
```
