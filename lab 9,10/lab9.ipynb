```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
y = np.array([40, 45, 50, 55, 65, 70])

# Model
model = LinearRegression()
model.fit(X, y)

# Predictions
pred = model.predict(X)

print("Slope: ", model.coef_)
print("Intercept: ", model.intercept_)

# Test with new value
test_value = np.array([[7]])   # 7 study hours
prediction = model.predict(test_value)



# Plot
plt.scatter(X, y, label="Actual Data")
plt.plot(X, pred, label="Best Fit Line")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.show()


print("Predicted marks for 7 hours:", prediction)
```

```python
hours = np.array([1, 2, 3, 4, 5, 6])
marks = np.array([45, 50, 55, 60, 65,70])


h_mean = np.mean(hours)
m_mean = np.mean(marks)

m = np.sum((hours - h_mean) * (marks - m_mean)) / np.sum((hours - h_mean)**2) # m = ( (hours - hours_mean) * (marks - marks_mean) ) / (hours - hours_mean)**2


b = m_mean - h_mean * m # b = y - mx

print("Slope (m):", m)
print("Intercept (b):", b)


new_hours = 7
predicted_marks = m * new_hours + b #y = mx + b
print(f"Predicted Marks for {new_hours} study hours for the KING UMAIR GUL:", predicted_marks)
```
