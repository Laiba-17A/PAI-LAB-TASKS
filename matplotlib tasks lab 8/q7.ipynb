```python
import matplotlib.pyplot as plt
import numpy as np

years = np.arange(1925, 2025)
sea_level = 0.02 * (years - 1925) + np.random.normal(0, 0.1, len(years))

plt.scatter(years, sea_level)
plt.xlabel("year")
plt.ylabel("sea level (m)")
plt.title("sea level rise (100 years)")
plt.show()
```
