```python
import matplotlib.pyplot as plt

names = [f"S{i}" for i in range(1, 21)]
heights = [150,155,160,165,152,158,170,175,168,162,159,173,180,177,169,166,172,174,171,163]

colors = plt.cm.tab20(range(20))

plt.bar(names, heights, color=colors)
plt.title("Student Heights (Bar Chart)")
plt.xlabel("Students")
plt.ylabel("Height (cm)")
plt.show()

plt.pie(heights, labels=names, autopct="%1.1f%%")
plt.title("Student Heights (Pie Chart)")
plt.show()
```
