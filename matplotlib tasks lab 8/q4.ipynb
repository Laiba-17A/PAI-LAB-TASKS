```python
import matplotlib.pyplot as plt

ages = [19,20,22,24,26,28,30,33,35,37,40,18,23,29,32]

groups = ["18-25","26-30","31-35","36+"]

g1 = sum(18 <= a <= 25 for a in ages)
g2 = sum(26 <= a <= 30 for a in ages)
g3 = sum(31 <= a <= 35 for a in ages)
g4 = sum(a >= 36 for a in ages)

sizes = [g1, g2, g3, g4]

plt.pie(sizes, labels=groups, autopct="%1.1f%%")
plt.title("age distribution")
plt.show()
```
