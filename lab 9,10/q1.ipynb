```python
import numpy as np

X = np.array([1.1,1.3,1.5,2.0,2.2,2.9,3.0,3.2,3.2,3.7])
y = np.array([39,46,47,52,56,64,65,67,68,70])

X_mean = np.mean(X)
y_mean = np.mean(y)

m = np.sum((X - X_mean)*(y - y_mean)) / np.sum((X - X_mean)**2)

b = y_mean - m*X_mean

print("slope (m):", m)
print("intercept (b):", b)

X_new = 4.5
y_pred = m*X_new + b
print("predicted salary for 4.5 years:", y_pred)

y_hat = m*X + b
totalsumsq = np.sum((y - y_mean)**2)
residsumsq = np.sum((y - y_hat)**2)
R2 = 1 - residsumsq/totalsumsq
print("R² value:", R2)
```

```python
from sklearn.linear_model import LinearRegression

X_reshaped = X.reshape(-1,1)

model = LinearRegression()
model.fit(X_reshaped, y)

print("slope (m):", model.coef_[0])
print("intercept (b):", model.intercept_)

y_pred_skl = model.predict([[4.5]])
print("predicted salary for 4.5 years:", y_pred_skl[0])

# Compute R² using sklearn
R2_skl = model.score(X_reshaped, y)
print("R² :", R2_skl)
```
