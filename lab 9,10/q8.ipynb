```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

house_data = {
    'QualityScore': [6, 7, 5, 8, 7, 9, 4, 6, 8, 5],
    'LivingArea': [1500, 1850, 1200, 2100, 1750, 2400, 980, 1600, 1950, 1100],
    'GarageCapacity': [2, 3, 1, 2, 2, 3, 1, 2, 3, 1],
    'ConstructionYr': [1998, 2005, 1970, 2010, 1985, 2007, 1963, 1995, 2001, 1980],
    'Price': [180000, 235000, 140000, 260000, 200000, 300000, 120000, 175000, 245000, 150000]
}

df = pd.DataFrame(house_data)

predictors = ['QualityScore', 'LivingArea', 'GarageCapacity', 'ConstructionYr']
target = 'Price'

X = df[predictors]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("\n=== Regression Coefficients ===")
for name, value in zip(predictors, model.coef_):
    print(f"{name:15} -> {value:.3f}")

print("\nIntercept:", round(model.intercept_, 3))
print("R² Score :", r2_score(y_test, y_pred))
print("RMSE     :", np.sqrt(mean_squared_error(y_test, y_pred)))

impact_rank = sorted(zip(predictors, np.abs(model.coef_)), key=lambda x: -x[1])
print("\nMost Influential Feature:", impact_rank[0])
```
