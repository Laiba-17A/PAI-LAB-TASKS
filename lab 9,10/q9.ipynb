```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error

data = {
    'RAM_GB': [8, 16, 4, 8, 32, 16, 8, 12, 4, 32],
    'Weight_kg': [1.8, 2.1, 1.5, 1.7, 2.4, 2.0, 1.6, 1.9, 1.4, 2.3],
    'CPU_GHz': [2.5, 3.2, 1.8, 2.2, 3.5, 2.8, 2.4, 3.0, 1.6, 3.4],
    'Storage_GB': [256, 512, 128, 256, 1024, 512, 256, 512, 128, 1024],
    'Price_USD': [700, 1200, 500, 750, 2000, 1300, 800, 1100, 450, 1900]
}

df = pd.DataFrame(data)

features = ['RAM_GB', 'Weight_kg', 'CPU_GHz', 'Storage_GB']
target = 'Price_USD'

X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=15)

model_raw = LinearRegression()
model_raw.fit(X_train, y_train)
pred_raw = model_raw.predict(X_test)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model_scaled = LinearRegression()
model_scaled.fit(X_train_scaled, y_train)
pred_scaled = model_scaled.predict(X_test_scaled)

print("Unscaled Model Performance")
print("R2:", r2_score(y_test, pred_raw))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_raw)))

print("\nScaled Model Performance")
print("R2:", r2_score(y_test, pred_scaled))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred_scaled)))

coef_impact = sorted(zip(features, np.abs(model_scaled.coef_)), key=lambda x: -x[1])
print("\nMost Significant Predictor:", coef_impact[0])
```
